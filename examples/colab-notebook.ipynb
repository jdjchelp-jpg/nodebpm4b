# BPM4B - Professional Multimedia Converter in Google Colab

This notebook demonstrates how to use BPM4B in Google Colab with automatic tunneling for remote access.

## Features
- **5x Faster Processing**: Parallel audio processing with configurable concurrency
- **Automatic Tunneling**: Exposes localhost to the internet via localtunnel or ngrok
- **Colab Optimizations**: Automatically detects and applies Colab-specific optimizations
- **All Formats**: Convert MP3, WAV, FLAC, AAC, OGG, OPUS, M4A, M4B, MP4, MKV, AVI, MOV, WEBM, and more
- **Folder to M4B**: Convert entire folders to audiobooks with automatic chapter markers

## Step 2: Start the Server with Automatic Tunneling

````python
import subprocess
import time
import threading
import requests
from IPython.display import HTML, display

# Start BPM4B server in background with tunneling enabled
server_process = subprocess.Popen(
    ['bpm4b', 'web', '--port', '5000'],
    stdout=subprocess.PIPE,
    stderr=subprocess.PIPE,
    text=True
)

# Wait for server to start
time.sleep(5)

print("✅ BPM4B server started on port 5000")
print("⏳ Starting tunnel for remote access...")
````
